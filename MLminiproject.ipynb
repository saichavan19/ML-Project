```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, mean_squared_error

# Load Dataset

df = pd.read_csv("student_data.csv")

# Convert categorical to numeric
df['Gender'] = df['Gender'].map({'Male':0, 'Female':1})
df['Internet_Usage'] = df['Internet_Usage'].map({'Low':0,'Medium':1,'High':2})
df['Extra_Activities'] = df['Extra_Activities'].map({'No':0,'Yes':1})
df['Placed'] = df['Placed'].map({'No':0,'Yes':1})

# Regression (Final Score)

X_reg = df[['Study_Hours','Attendance','Internal_Marks']]
y_reg = df['Final_Exam_Score']

X_train, X_test, y_train, y_test = train_test_split(X_reg, y_reg, test_size=0.2)

# Simple Linear Regression
slr = LinearRegression()
slr.fit(X_train[['Study_Hours']], y_train)
pred_slr = slr.predict(X_test[['Study_Hours']])

# Multiple Regression
mlr = LinearRegression()
mlr.fit(X_train, y_train)
pred_mlr = mlr.predict(X_test)

print("SLR MSE:", mean_squared_error(y_test, pred_slr))
print("MLR MSE:", mean_squared_error(y_test, pred_mlr))

# Classification (Placement)

X_clf = df.drop(['Placed','Final_Exam_Score'], axis=1)
y_clf = df['Placed']

X_train, X_test, y_train, y_test = train_test_split(X_clf, y_clf, test_size=0.2)

# Decision Tree
dt = DecisionTreeClassifier()
dt.fit(X_train, y_train)
print("DT Accuracy:", accuracy_score(y_test, dt.predict(X_test)))

# Logistic Regression
lr = LogisticRegression()
lr.fit(X_train, y_train)
print("LR Accuracy:", accuracy_score(y_test, lr.predict(X_test)))

# SVM
svm = SVC()
svm.fit(X_train, y_train)
print("SVM Accuracy:", accuracy_score(y_test, svm.predict(X_test)))

# KNN
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
print("KNN Accuracy:", accuracy_score(y_test, knn.predict(X_test)))

# Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
print("NB Accuracy:", accuracy_score(y_test, nb.predict(X_test)))
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

# Convert data to binary (example)
df_ap = df[['Study_Hours','Attendance']]
df_ap = df_ap.applymap(lambda x: 1 if x > df_ap.mean().mean() else 0)

# Generate frequent itemsets using apriori algorithm
freq_items = apriori(df_ap, min_support=0.2, use_colnames=True)

# Generate association rules from frequent itemsets
rules = association_rules(freq_items, metric="confidence", min_threshold=0.5)

print(rules)
```
